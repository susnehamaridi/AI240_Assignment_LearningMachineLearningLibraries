```python
import pandas as pd
import numpy as np

data = {
    'Student': ['Anu', 'Ravi', 'Priya', 'John', 'Sneha'],
    'Quiz1': [85, np.nan, 91, 68, np.nan],
    'Quiz2': [90, 78, np.nan, 75, 93]
}

df = pd.DataFrame(data)

print(df)
```

```python
cleaned_drop = df.dropna()

print(cleaned_drop)
```

```python
cleaned_fill = df.copy()

cleaned_fill['Quiz1'] = cleaned_fill['Quiz1'].fillna(
    cleaned_fill['Quiz1'].mean()
)

cleaned_fill['Quiz2'] = cleaned_fill['Quiz2'].fillna(
    cleaned_fill['Quiz2'].mean()
)

print(cleaned_fill)
```

```python
print("Original rows:", len(df))
print("Rows after dropna():", len(cleaned_drop))
print("Rows after fillna():", len(cleaned_fill))
```

```python
import pandas as pd

players = pd.DataFrame({
    'Name': ['LeBron James', 'Stephen Curry', 'Kevin Durant',
             'Chris Paul', 'Anthony Davis'],
    'Age': [41, 38, 37, 41, 33],
    'Salary': [48000000, 56000000, 51000000, 3600000, 54000000],
    'Team': ['Lakers', 'Warriors', 'Rockets', 'Clippers', 'Mavericks']
})

print(players)
```

```python
print(players.iloc[:4])
```

```python
print(players[['Name', 'Age', 'Salary']])
```

```python
print(players.iloc[1, 2])
```

```python
older_players = players[players['Age'] > 34]

print(older_players)
```

```python
team_data = pd.DataFrame({
    'Name': ['LeBron James', 'Stephen Curry', 'Kevin Durant',
             'Chris Paul', 'Anthony Davis'],
    'Team': ['Lakers', 'Warriors', 'Rockets', 'Clippers', 'Mavericks']
})

# Remove Team from the original DataFrame to avoid duplicate columns
players_basic = players.drop(columns=['Team'])

# Merge the two DataFrames using Name
final_df = pd.merge(players_basic, team_data, on='Name', how='left')

print(final_df)
```

```python
final_df.to_csv('players_final.csv', index=False)
```
