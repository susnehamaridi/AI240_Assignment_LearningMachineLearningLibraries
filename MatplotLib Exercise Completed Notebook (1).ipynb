## Exercise 2 - Import Matplotlib using

import matplotlib.pyplot as plt



```python
import matplotlib.pyplot as plt
import numpy as np
```

## Exercise 3 - Draw Barchart




```python
# Data from slide 12
x = np.array(["A", "B", "C", "D"])
y = np.array([5, 7, 3, 11])

plt.figure(figsize=(7, 5))
plt.bar(x, y, color="skyblue", edgecolor="black")

plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")

plt.tight_layout()
plt.show()
```

## Exercise 4 -  Draw Histogram


```python
# Data-generation method from slide 13
x = np.random.normal(170, 10, 250)

plt.figure(figsize=(7, 5))
plt.hist(x, bins=10, color="lightgreen", edgecolor="black")

plt.title("Histogram of Normally Distributed Values")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## Exercise 5 -  Draw Piechart



```python
# Data from slide 15
y = np.array([35, 25, 25, 15])

labels = ["Category A", "Category B", "Category C", "Category D"]

plt.figure(figsize=(7, 7))
plt.pie(
    y,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=["skyblue", "lightgreen", "gold", "salmon"]
)

plt.title("Pie Chart")
plt.axis("equal")

plt.tight_layout()
plt.show()
```
