```python
measurement_colors = [
    '#78C850',  # Sepal length
    '#F08030',  # Sepal width
    '#6890F0',  # Petal length
    '#A8B820'   # Petal width
]

plt.figure(figsize=(10, 8))
sns.violinplot(data=iris, palette=measurement_colors)
plt.title("Violin Plot of Iris Measurements")
plt.ylabel("Measurement (cm)")
plt.show()
```

# Exercise 6 - Draw a Swarm Plot

A swarm plot is another way of plotting the distribution of an attribute or the joint distribution of a couple of attributes.

Create a Swarm plot of Petal length across the species.
 
sns.swarmplot(y='petal_length', x='species', data=iris, palette=species_colors)

```python
species_colors = ['#78C850', '#F08030', '#6890F0']

plt.figure(figsize=(10, 8))

sns.swarmplot(
    data=iris,
    x='species',
    y='petal_length',
    hue='species',
    palette=species_colors,
    legend=False
)

plt.title("Petal Length Across Iris Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")

plt.show()
```

# Exercise 8 - Draw a Heatmap Plot

corrmat = iris.corr()

sns.heatmap(corrmat, annot = True, square = True)


```python

import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

iris = pd.DataFrame(
    load_iris().data,
    columns=load_iris().feature_names
)

corrmat = iris.corr()

sns.heatmap(corrmat, annot=True, square=True)
plt.show()
```
